```python
import rasterio
from rasterio.windows import Window

dataset = rasterio.open("sample-data/sample_raster.tif")
print(dataset.bounds)
```

```python
window = Window(100, 100, 200, 200)

data = dataset.read(1, window=window)
transform = dataset.window_transform(window)

with rasterio.open(
    "sample-data/cropped_with_location.tif",
    "w",
    driver="GTiff",
    height=data.shape[0],
    width=data.shape[1],
    count=1,
    dtype=data.dtype,
    crs=dataset.crs,
    transform=transform,
) as new_file:
    new_file.write(data, 1)

print("saved")
```

```python
cropped = rasterio.open("sample-data/cropped_with_location.tif")
print(cropped.bounds)
print(cropped.crs)
```
