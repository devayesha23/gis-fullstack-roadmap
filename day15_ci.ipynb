```python
import os
os.makedirs(".github/workflows", exist_ok=True)
```

```python
%%writefile .github/workflows/tests.yml
name: tests
on:
  push:
  pull_request:
jobs:
  pytest:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v4
      - uses: actions/setup-python@v5
        with:
          python-version: "3.11"
      - run: pip install geopandas rasterio rasterstats shapely numpy sqlalchemy fastapi httpx pytest
      - run: python -m pytest -q
```

```python
from pathlib import Path
p = Path("README.md")
s = p.read_text(encoding="utf-8")
badge = "![tests](https://github.com/devayesha23/gis-fullstack-roadmap/actions/workflows/tests.yml/badge.svg)"
if "badge.svg" not in s:
    s = s.replace("# GIS Full-Stack Roadmap\n", "# GIS Full-Stack Roadmap\n\n" + badge + "\n", 1)
    p.write_text(s, encoding="utf-8")
print("badge.svg" in s)
```
